## Determine cost thresholds for simulated annealing

1/11/2025

Matlab has default values for convergence. Python's scikitlearn libraries
do not. So we specify 5,000 iterations by default; we use 10,000 to get a
better fit. Python will exhaust all of those regardless of convergence,
unless we implement a callback function to indicate adequate convergence.

In this notebook, we review six successful STARE runs from three scanners
to get a sense for the range of values from the cost function and where
we might want to stop the annealer early to save processing time.

```python
%load_ext autoreload
%autoreload 2

from pathlib import Path
import pickle
```

```python
output_base_path = Path("/mnt/brunodata/pet")
```

```python
results_files = dict()
for results_path in output_base_path.glob("*/derivatives/complete_0105_nosr/clip-0/*"):
    scanner = results_path.parent.parent.parent.parent.name
    subject_id = results_path.name
    results_file = results_path / "cache" / f"sub-{subject_id}_step-4_results.pkl"
    results_str = "has results" if results_file.exists() else "    ...    "
    html_file = results_path / f"sub-{subject_id}.html"
    complete_str = "and html" if html_file.exists() else "..."
    print(f"Subject {subject_id} ({scanner}) {results_str} {complete_str}.")

    if results_file.exists():
        results_files[subject_id] = {
            "scanner": scanner,
            "results_file": results_file,
            "html_file": html_file,
        }
```

```python
for subject_id, subject_dict in results_files.items():
    if subject_dict["results_file"].exists():
        results = pickle.load(open(subject_dict["results_file"], "rb"))
    subject_dict["sa_results"] = results.annealer_results
```
